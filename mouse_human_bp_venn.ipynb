```python
import matplotlib.pyplot as plt
from matplotlib_venn import venn2
import seaborn as sns

# Set up the data
mouse_bp = 64685173
human_bp = 63417540

shared_bp = 23153655
mouse_hg_pct = 1.97
human_hg_pct = 1.92

# Calculate the unique values for each set
mouse_unique = mouse_bp - shared_bp
human_unique = human_bp - shared_bp

# Calculate percentages
mouse_percent = (mouse_unique / mouse_bp) * 100
human_percent = (human_unique / human_bp) * 100
shared_percent_mouse = (shared_bp / mouse_bp) * 100
shared_percent_human = (shared_bp / human_bp) * 100

# Format numbers with commas
mouse_formatted = f"{mouse_bp:,}"
human_formatted = f"{human_bp:,}"
shared_formatted = f"{shared_bp:,}"

# Create the Venn diagram
fig = plt.figure(figsize=(10, 6), dpi=300)
venn = venn2(subsets=(mouse_unique, human_unique, shared_bp), 
             set_labels=(f'Mouse', f'Human'))

# Set font sizes for all text elements to match Jupyter display
label_fontsize = 15
venn_label_fontsize = 13
sub_label_fontsize = 9

m_text = f'\n{mouse_formatted}bp\n{mouse_hg_pct}% of human genome'
h_text = f'\n{human_formatted}bp\n{human_hg_pct}% of human genome'

# Adjust set label font sizes
for label in venn.set_labels:
    label.set_fontsize(label_fontsize)

# use venn.set_labels to get the position of mouse and put the m_text under it
x, y = venn.set_labels[0].get_position()
y -= 0.01
x -= 0.07
plt.text(x, y, m_text, ha='center', va='top', fontsize=sub_label_fontsize)

m_x, m_y = venn.set_labels[0].get_position()
h_x, h_y = venn.set_labels[1].get_position()
venn.set_labels[1].set_position((h_x, m_y))
x, y = venn.set_labels[1].get_position()
y -= 0.01
x += 0.09
plt.text(x, y, h_text, ha='center', va='top', fontsize=sub_label_fontsize)

# use colors from "Set3"
color_palette = sns.color_palette("pastel", 12)

# Customize the diagram
venn.get_patch_by_id('10').set_color(color_palette[-3])
venn.get_patch_by_id('01').set_color(color_palette[-7])
venn.get_patch_by_id('11').set_color(color_palette[-2])

# Add labels with numbers and percentages, and set font size
venn.get_label_by_id('10').set_text(f'{mouse_unique:,}bp\n({mouse_percent:.1f}%)')
venn.get_label_by_id('01').set_text(f'{human_unique:,}bp\n({human_percent:.1f}%)')
venn.get_label_by_id('11').set_text(f'{shared_formatted}bp')

for label_id in ['10', '01', '11']:
    label = venn.get_label_by_id(label_id)
    if label is not None:
        label.set_fontsize(venn_label_fontsize)

sub_mid_label = f'\n({shared_percent_mouse:.1f}% of Mouse)\n({shared_percent_human:.1f}% of Human)'
m_x, m_y = venn.get_label_by_id('11').get_position()

# y += 0.3
# x += 0.00
plt.text(m_x, m_y, sub_mid_label, ha='center', va='top', fontsize=sub_label_fontsize)


# Move the 11 label a bit upper
venn.get_label_by_id('11').set_y(venn.get_label_by_id('11').get_position()[1] + 0.15)
# Adjust label positions
venn.get_label_by_id('10').set_y(venn.get_label_by_id('10').get_position()[1] + 0.1)
venn.get_label_by_id('01').set_y(venn.get_label_by_id('01').get_position()[1] + 0.1)
venn.get_label_by_id('11').set_y(venn.get_label_by_id('11').get_position()[1] - 0.15)  # Adjusted to accommodate the extra line

venn.get_label_by_id('01').set_y(venn.get_label_by_id('11').get_position()[1])
venn.get_label_by_id('10').set_y(venn.get_label_by_id('11').get_position()[1])

# Save the figure as svg in high quality, ensuring font size is preserved
plt.savefig('results/Supp_Data_4_hg38_mouse_mm10_converted_w_human_sc/figures/venn_mouse_human_bp.svg', format='svg')
```
